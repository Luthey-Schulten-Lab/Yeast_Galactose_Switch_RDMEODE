```python
odeTs = odeTraj['ts']/60 # convert to minutes
# odeTs = odeTraj['ts']
NAV = 6.022e23*(traj.reg.cytoplasm.volume + traj.reg.nucleoplasm.volume + traj.reg.plasmaMembrane.volume)
odeYs = {n:odeTraj['ys'][:,i]*NAV for i,n in enumerate(odeTraj['names'])}
rdmeYs = dict()

for sp in traj.speciesList:
    ts,ys = sp.getNumberTrajectory()
    rdmeYs[sp.name] = ys
    rdmeTs = ts/60  # convert to minutes
```

Only show G2 related species

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Assuming rdmeYs, odeYs, rdmeTs, odeTs, and traj_fname are already defined

# Filter names to include only those with 'G2' or 'R2'
names = sorted(filter(lambda x: 'G2' in x or 'R2' in x, set(rdmeYs.keys()) | set(odeYs.keys())))
nplots = len(names)
ncols = 3  # Reduced number of columns for better visibility
nrows = int(np.ceil(nplots / ncols))

# Arguments for the axes
axargs = dict(xlim=(min(odeTs.min(), rdmeTs.min()), max(odeTs.max(), rdmeTs.max())),
              xlabel='t/min', ylabel='Count', xticks=np.arange(0, 61, 10))

# Create figure and subplots
fig, axs = plt.subplots(ncols=ncols, nrows=nrows, figsize=(15, 4 * nrows))

# Adjust layout to make space for the suptitle
fig.subplots_adjust(top=0.95)

# Add a title to the figure without overlapping with subplots
fig.suptitle(f"G2 and R2 species - {traj_file}", y=0.98)

axs = axs.ravel()
colors = sns.color_palette()

# Plot data
for ax, name in zip(axs, names):
    if name in rdmeYs:
        rdmeHandle = ax.plot(rdmeTs, rdmeYs[name], c=colors[0], label="RDME")[0]
    if name in odeYs:
        odeHandle = ax.plot(odeTs, odeYs[name], c=colors[1], label="ODE")[0]
    ax.set(title=name, **axargs)

# Remove empty subplots
for i in range(nplots, len(axs)):
    fig.delaxes(axs[i])

# Tight layout to adjust subplots spacing
fig.tight_layout(rect=[0, 0.03, 1, 0.95])

# Add legend to the figure
fig.legend([rdmeHandle, odeHandle], ["RDME", "ODE"], 
           loc='lower center', ncol=2, bbox_to_anchor=(0.5, 0))

# Save the figure
fig.savefig(traj_dir + traj_file + '_G2_R2_species_counts.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python
# Demonstrating the duration of DG2 in different states

# Extract relevant data
ode_DG2 = rdmeYs['DG2']
ode_DG2_G4d = rdmeYs['DG2_G4d']
ode_DG2_G4d_G80d = rdmeYs['DG2_G4d_G80d']

# Create a figure and axis
fig, ax = plt.subplots(figsize=(10, 6))

# Plot stacked area chart
ax.fill_between(rdmeTs, 0, ode_DG2, alpha=0.5, label='DG2')
ax.fill_between(rdmeTs, ode_DG2, ode_DG2 + ode_DG2_G4d, alpha=0.5, label='DG2_G4d')
ax.fill_between(rdmeTs, ode_DG2 + ode_DG2_G4d, ode_DG2 + ode_DG2_G4d + ode_DG2_G4d_G80d, alpha=0.5, label='DG2_G4d_G80d')

# Customize the plot
ax.set_xlabel('Time (min)')
ax.set_ylabel('Count')
ax.set_title(f'Duration of DG2 in Different States - {traj_file}')
ax.legend(fontsize='small')  # Reduce legend font size

# Add grid for better readability
ax.grid(True, linestyle='--', alpha=0.7)

# Adjust layout and display
plt.tight_layout()
plt.show()

# Calculate and print the average time spent in each state
total_time = rdmeTs[-1] - rdmeTs[0]
avg_time_DG2 = np.trapz(ode_DG2, rdmeTs) / total_time
avg_time_DG2_G4d = np.trapz(ode_DG2_G4d, rdmeTs) / total_time
avg_time_DG2_G4d_G80d = np.trapz(ode_DG2_G4d_G80d, rdmeTs) / total_time

print(f"Average time spent in DG2 state: {avg_time_DG2:.2f} units")
print(f"Average time spent in DG2_G4d state: {avg_time_DG2_G4d:.2f} units")
print(f"Average time spent in DG2_G4d_G80d state: {avg_time_DG2_G4d_G80d:.2f} units")

# Save the figure
fig.savefig(traj_dir + traj_file + '_DG2_states.png', dpi=300, bbox_inches='tight')
```
